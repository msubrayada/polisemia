```python
import pandas as pd
```

```python
def cargar_datos(model_name):
    data_og = pd.read_csv(f"output/tokens_{model_name}_state_twitter_2500_2021--10--06.csv")
    last_id_tweet = int(data_og.tail(1)['id_tweet'])
    d = pd.read_csv(f"output/tokens_{model_name}_state_twitter_2501_2021--10--11.csv")
    d['id_tweet'] += last_id_tweet
    data_og = pd.concat([data_og, d], ignore_index=True)
    last_id_tweet = int(data_og.tail(1)['id_tweet'])
    d = pd.read_csv(f"output/tokens_{model_name}_state_twitter_2502_2021--10--12.csv")
    #d = d[27081:] ### quitar
    #d['id_tweet'] = d['id_tweet'].astype(int) ### quitar
    d['id_tweet'] += last_id_tweet
    data_og = pd.concat([data_og, d], ignore_index=True)
    data_og['token'] = data_og['token'].astype(str)
    return data_og
```

```python
model_names = ['beto', 'bertin', 'robertuito', 'mex_state', 'mex_large']
for m in model_names:
    datos = cargar_datos(m)
    datos.to_csv(f"output/tokens_{m}.csv", index=False)
```
