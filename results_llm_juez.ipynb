```python
import pandas as pd
import glob
import json
from collections import Counter
```

```python
def get_tweet(palabras_df, id_tweet, special_token=None):

    RED = "\033[91m"
    RESET = "\033[0m"


    l = list(palabras_df[palabras_df['id_tweet']==id_tweet]['texto'])
    l = l[0].split()
    l = [str(_) for _ in l]

    if special_token==None:
        print(" ".join(l))
        return

    s = ""
    for w in l:
        if w==special_token:
            s += RED + w + RESET + " "

        else:
            s += w + " "

    return s
```

```python
path = "drive/MyDrive/infotec/polisemia/"
path = './'
textos_df = pd.read_csv(path + 'output/all_tweets.csv')
```

```python
model = "qwen_model/"
files = glob.glob(path + "juicios_llm_" + model.replace("/", "_") + "/*")
files
```

```python
total_tweets = 0
tweets_usables = 0

for f in files[0:1]:
    data = json.load(open(f, encoding='utf-8'))
    total_tweets += len(data['asignaciones'])
    counter = Counter([a['sentido'] for a in data['asignaciones']])
    num_clusters = 0
    t_u = 0
    for k in counter:        
        #if k=="no_claro":
        #    continue
        c = counter[k]
        if c >= 5:
            t_u += c
            num_clusters += 1
    if num_clusters > 1:
        tweets_usables += t_u
total_tweets, tweets_usables, total_tweets - tweets_usables
```
