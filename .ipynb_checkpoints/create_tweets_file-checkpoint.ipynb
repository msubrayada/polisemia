```python
import pandas as pd
import glob
```

```python
path = "d:/data/mexican_test"
path_output = "output/"
files = glob.glob(path + "/*.txt")
```

```python
content = []
for num_file, file_path in enumerate(files[0:3]):

    with open(file_path, "r", encoding="utf-8") as file:
        text = [line.strip() for line in file if line.strip()]
    content += text
```

```python
len(content)
```

```python
df = pd.DataFrame(content, columns=['texto'])

df = df.rename_axis("id_tweet")
df.to_csv("output/all_tweets.csv")
```
