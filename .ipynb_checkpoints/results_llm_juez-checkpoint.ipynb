```python
import pandas as pd
import glob
import json
```

```python
def get_tweet(palabras_df, id_tweet, special_token=None):

    RED = "\033[91m"
    RESET = "\033[0m"


    l = list(palabras_df[palabras_df['id_tweet']==id_tweet]['texto'])
    l = l[0].split()
    l = [str(_) for _ in l]

    if special_token==None:
        print(" ".join(l))
        return

    s = ""
    for w in l:
        if w==special_token:
            s += RED + w + RESET + " "

        else:
            s += w + " "

    return s
```

```python
path = "drive/MyDrive/infotec/polisemia/"
textos_df = pd.read_csv(path + 'output/all_tweets.csv')
```

```python
model = "qwen_model/"
files = glob.glob(path + "juicios_llm_" + model.replace("/", "_") + "/*")
files
```

```python
for f in files:
    print('-'*30)
    data = json.load(open(f))
    print(data['palabra'])
    for s in data['sentidos']:
        print(s['id'], s['glosa'])
    print()

    # Sort data in order of key 'sentido'
    sorted_asignaciones = sorted(data['asignaciones'], key=lambda x: x['sentido'])

    for a in sorted_asignaciones:
        #print(a['sentido'], textos_df[textos_df['id_tweet'] == int(a['id_tweet'])]['texto'].to_list())
        print(a['sentido'], get_tweet(textos_df, int(a['id_tweet']), special_token=data['palabra']))
```
