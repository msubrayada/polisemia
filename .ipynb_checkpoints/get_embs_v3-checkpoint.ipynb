```python
# %% [markdown]
# ## 1. Cargar la lista de palabras objetivo (intersección de vocabulario)
 
# %%
interseccion = pd.read_csv(INTERSECCION_CSV)
interseccion = interseccion[~interseccion["word_normalized"].isin(stopwords)]
interseccion = interseccion[interseccion["word_normalized"].str.len()>3]
interseccion = interseccion[interseccion["word_normalized"].str.isalpha()]
palabras_objetivo = set(interseccion["word_normalized"])

print(f"Palabras objetivo: {len(palabras_objetivo)}")

def cargar_textos(ruta=TEXTOS_CSV):
    df = pd.read_csv(ruta)        
    #df['texto'] = df['texto'].str.split('_GEO ', expand=True)[1]
    return dict(zip(df["id_tweet"], df["texto"]))
 
textos_por_tweet = cargar_textos()
print(f"Tweets con texto cargado: {len(textos_por_tweet)}")
```

```python
# %% [markdown]
# ## 2. Funciones de extracción
 
# %%

def encontrar_ocurrencias_regex(palabra, texto):
    """Todas las ocurrencias de `palabra` como palabra completa (sin distinguir
    mayúsculas/minúsculas) en `texto`. Devuelve lista de (char_start, char_end)."""
    patron = r"\b" + re.escape(palabra) + r"\b"
    return [(m.start(), m.end()) for m in re.finditer(patron, texto, flags=re.IGNORECASE)]
    
"""
def construir_secuencias_por_tweet(tokens_path):

    df = pd.read_csv(tokens_path, keep_default_na=False)
    df = df.sort_values(["id_tweet", "id_vector"])
    secuencias = {}
    for id_tweet, grupo in df.groupby("id_tweet"):
        secuencias[id_tweet] = list(zip(grupo["id_vector"].tolist(), grupo["token"].tolist()))
    
        
    return secuencias 
"""
 
 
def extraer_embeddings_modelo(nombre_modelo, hf_id, palabras_path, textos_por_tweet, palabras_objetivo):
    """
    Tokeniza el texto CRUDO de cada tweet directamente (una sola llamada,
    con return_offsets_mapping=True) y localiza cada ocurrencia de palabra
    objetivo por posición de caracter (regex), no por reconstrucción de
    sub-tokens. Devuelve dos dicts alineados por índice:
      - embeddings_por_palabra: {palabra: [embeddings de cada ocurrencia]}
      - tweets_por_palabra: {palabra: [id_tweet de cada ocurrencia]}
    """
    print(f"\n--- {nombre_modelo} ---")    
        
    tokenizer = AutoTokenizer.from_pretrained(hf_id)
    if not tokenizer.is_fast:
        raise ValueError(f"{nombre_modelo}: se necesita un tokenizador 'fast' para usar offset_mapping")
    model = AutoModel.from_pretrained(hf_id, output_hidden_states=True).to(DEVICE).eval()
 
    palabras_df = pd.read_csv(palabras_path)
    palabras_df = palabras_df[palabras_df["word_normalized"].isin(palabras_objetivo)]
    # qué palabras buscar en cada tweet (deduplicado -- el CONTEO real de
    # ocurrencias sale de la búsqueda por regex sobre el texto, no de aquí)
    palabras_por_tweet = defaultdict(set)
    for _, row in palabras_df.iterrows():
        #palabras_por_tweet[row["id_tweet"]].add(row["word_normalized"])
        palabras_por_tweet[row["id_tweet"]].add(
        (row["word_normalized"], 
         #row["id_vector_start"], 
         #row["id_vector_end"], 
         row['region']
        )
    )
 
    embeddings_por_palabra = defaultdict(list)
    tweets_por_palabra = defaultdict(list)
    ids_tweets = [t for t in palabras_por_tweet.keys() if t in textos_por_tweet]
    print(f"Tweets a procesar: {len(ids_tweets)}")
 
    for i in range(0, len(ids_tweets), BATCH_SIZE):
        lote_ids = ids_tweets[i : i + BATCH_SIZE]
        if nombre_modelo in ['beto', 'robertuito', 'bertin']:
            lote_textos = [str(textos_por_tweet[tid].split('_GEO ')[-1]) for tid in lote_ids]
        else:
            lote_textos = [str(textos_por_tweet[tid]) for tid in lote_ids]
 
        encoded = tokenizer(
            lote_textos,
            padding=True,
            truncation=True,
            max_length=MAX_LENGTH,
            return_tensors="pt",
            return_offsets_mapping=True,
        )
        offsets_lote = encoded.pop("offset_mapping").tolist()  # (batch, seq_len, 2), en CPU
        encoded = encoded.to(DEVICE)
 
        with torch.no_grad():
            out = model(**encoded)
        hidden = out.hidden_states[CAPA_EXTRACCION]  # (batch, seq_len, hidden_dim)
 
        for b, tid in enumerate(lote_ids):
            texto = lote_textos[b]
            offsets = offsets_lote[b]  # lista de [char_start, char_end]; (0,0) = especial/padding
            for palabra, _ in palabras_por_tweet[tid]:
                for c_start, c_end in encontrar_ocurrencias_regex(palabra, texto):
                    idx_tokens = [
                        j for j, (s, e) in enumerate(offsets)
                        if not (s == 0 and e == 0) and s < c_end and e > c_start
                    ]
                    if not idx_tokens or max(idx_tokens) >= hidden.shape[1]:
                        continue  # sin tokens que cubran el span, o se truncó por max_length
                    vec = hidden[b, idx_tokens, :].mean(dim=0).cpu().numpy()
                    embeddings_por_palabra[palabra].append(vec)
                    tweets_por_palabra[palabra].append(tid)
 
        if i % (BATCH_SIZE * 50) == 0:
            print(f"  {i}/{len(ids_tweets)} tweets procesados")
 
    del model
    torch.cuda.empty_cache()
    return embeddings_por_palabra, tweets_por_palabra
```

```python
#  %% [markdown]
# ## 3. Sanity check
#
# Confirma que el span de caracteres encontrado por regex realmente cae sobre
# los tokens correctos, decodificando justo esos tokens.
 
# %%
def sanity_check(nombre_modelo, hf_id, palabra_ejemplo, textos_por_tweet, palabras_path):
    tokenizer = AutoTokenizer.from_pretrained(hf_id)
    palabras_df = pd.read_csv(palabras_path)
    fila = palabras_df[palabras_df["word_normalized"] == palabra_ejemplo].iloc[0]
    tid = fila["id_tweet"]
    texto = str(textos_por_tweet[tid])
    print(f"--- {nombre_modelo} ---")
    print("Texto original:", texto)
 
    ocurrencias = encontrar_ocurrencias_regex(palabra_ejemplo, texto)
    print(f"Ocurrencias de '{palabra_ejemplo}' encontradas por regex: {ocurrencias}")
 
    encoded = tokenizer(texto, return_offsets_mapping=True, return_tensors="pt")
    offsets = encoded["offset_mapping"][0].tolist()
    tokens = tokenizer.convert_ids_to_tokens(encoded["input_ids"][0])
 
    for c_start, c_end in ocurrencias:
        idx_tokens = [j for j, (s, e) in enumerate(offsets) if not (s == 0 and e == 0) and s < c_end and e > c_start]
        print(f"  Span [{c_start},{c_end}) -> tokens: {[tokens[j] for j in idx_tokens]} (deberían formar '{palabra_ejemplo}')")
 
# Ejemplo de uso — descomenta y ajusta:
# sanity_check("beto", MODELOS["beto"], "banco", textos_por_tweet, PALABRAS_CSV["beto"])
```

```python
# %% [markdown]
# ## 4. Guardar embeddings (a nivel de palabra, antes de clusterizar)
#
# Guarda un .npz por modelo con dos arrays alineados por fila:
# `embeddings` (n_ocurrencias, dim) y `words` (n_ocurrencias,) — el
# word_normalized correspondiente a cada fila. Es la unidad de las 13,772
# palabras de la intersección (post mean-pooling), no sub-tokens individuales.
 
# %%

 
def guardar_embeddings_modelo(nombre_modelo, embeddings_por_palabra, tweets_por_palabra, carpeta_salida=CARPETA_EMBEDDINGS):
    os.makedirs(carpeta_salida, exist_ok=True)
    palabras_lista = []
    tweets_lista = []
    vectores = []
    for palabra, embs in embeddings_por_palabra.items():
        for v, tid in zip(embs, tweets_por_palabra[palabra]):
            palabras_lista.append(palabra)
            tweets_lista.append(tid)
            vectores.append(v)
    vectores = np.vstack(vectores)
    palabras_arr = np.array(palabras_lista)
    tweets_arr = np.array(tweets_lista)
    ruta = os.path.join(carpeta_salida, f"{nombre_modelo}_embeddings.npz")
    np.savez_compressed(ruta, embeddings=vectores, words=palabras_arr, id_tweets=tweets_arr)
    print(f"  Guardado: {ruta} ({vectores.shape[0]} ocurrencias, dim={vectores.shape[1]})")
    return ruta

def cargar_embeddings(nombre_modelo, carpeta_salida='embeddings'):
    data = np.load(f"embeddings/{nombre_modelo}_embeddings.npz", allow_pickle=True)
    X, words, id_t = data["embeddings"], data["words"], data['id_tweets']
    return X, words, id_t

def cargar_tweets_por_palabra(nombre_modelo, carpeta_salida='embeddings'):
    with open(f"embeddings/{nombre_modelo}_tweets_por_palabra.json", 'r') as file:
        data = json.load(file)
    return data
```

```python
# %% [markdown]
# ## 5. Orquestación — extraer y guardar para los 5 modelos
#
# Esto SOBREESCRIBE los `.npz` anteriores (construidos con el método de
# reconstrucción de sub-tokens, ahora descartado). Corre esto una vez (por
# modelo puede tardar según tu GPU).
 
# %%
for nombre_modelo, hf_id in MODELOS.items():
    if nombre_modelo == 'mex_state':
        continue
    embeddings_por_palabra, tweets_por_palabra = extraer_embeddings_modelo(
        nombre_modelo, hf_id, PALABRAS_CSV[nombre_modelo], textos_por_tweet, palabras_objetivo
    )
    guardar_embeddings_modelo(nombre_modelo, embeddings_por_palabra, tweets_por_palabra)
 
print("\nListo. Los .npz están en la carpeta 'embeddings/' — continúa en clusterizar_sentidos.py")
```
